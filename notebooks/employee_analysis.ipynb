# Employee Salary & Performance Analysis

## 1. Dataset Loading

This project analyzes employee salary and performance data to identify patterns and relationships between employee experience, salary, department, and performance.

```python
import pandas as pd

df = pd.read_csv("../dataset/employee_salary_performance.csv")

print("Rows and Columns:", df.shape)
df.head()
```

## 2. Data Inspection

In this step, the dataset is inspected to understand its structure, data types, missing values, and duplicate records.

```python
print("Data Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

## 3. Data Cleaning

The dataset was checked for missing values, duplicate records, and incorrect data types. No missing values or duplicate records were found. The data types are appropriate for analysis, so no major cleaning was required.

```python
# Check and remove duplicate records
df = df.drop_duplicates()

# Check missing values
print("Total Missing Values:", df.isnull().sum().sum())

# Check duplicate records
print("Total Duplicate Rows:", df.duplicated().sum())

print("\nDataset after cleaning:", df.shape)
```

## 4. Exploratory Data Analysis (EDA)

Exploratory Data Analysis is performed to understand the distribution of employee salary, experience, and performance. Basic statistics and correlations are used to identify important patterns in the dataset.

```python
df.describe()
```

### Salary and Performance Analysis

The following analysis examines average salary and performance rating across different departments.

```python
print("Average Salary:", round(df["Salary"].mean(), 2))
print("Average Performance Rating:", round(df["Performance_Rating"].mean(), 2))

print("\nAverage Salary by Department:")
print(df.groupby("Department")["Salary"].mean().round(2))

print("\nAverage Performance by Department:")
print(df.groupby("Department")["Performance_Rating"].mean().round(2))
```

### Correlation Analysis

Correlation is used to understand the relationship between numerical variables such as age, experience, salary, and performance rating.

```python
df[["Age", "Experience_Years", "Salary", "Performance_Rating"]].corr().round(2)
```

## 5. Data Visualization

Five visualizations are created to identify salary, experience, department, and performance patterns in the employee dataset.

1 — Salary by Department

```python
import matplotlib.pyplot as plt

df.groupby("Department")["Salary"].mean().sort_values().plot(kind="bar")

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
2 — Employees by Department
```

```python
df["Department"].value_counts().plot(kind="bar")

plt.title("Number of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

3 — Experience vs Salary

```python
plt.scatter(df["Experience_Years"], df["Salary"])

plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.tight_layout()
plt.show()
```

4 — Performance Rating Distribution

```python
df["Performance_Rating"].value_counts().sort_index().plot(kind="bar")

plt.title("Employee Performance Rating Distribution")
plt.xlabel("Performance Rating")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

5 — Salary Distribution

```python
plt.hist(df["Salary"], bins=10)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

## 6. Business Insights

Based on the exploratory data analysis and Power BI dashboard, the following key insights were identified:

1. The IT department has the highest average salary, while Operations has the lowest.
2. Employees with higher experience generally earn higher salaries.
3. Most employees have a performance rating of 3 or 4.
4. Software Engineer and Data Analyst roles have relatively higher average salaries.
5. Salary has a strong positive relationship with experience, while performance rating has very little relationship with salary.
